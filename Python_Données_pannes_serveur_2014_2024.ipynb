```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger les données depuis un fichier Excel
file_path = "C:/Users/Mediamonster/Downloads/Forment_Fotso_Modelisation_OR_Python/Données_pannes_serveur_2014_2024.xlsx"  # Remplacez par le chemin de votre fichier
data_pannes = pd.read_excel(file_path)

# Filtrer les données pour exclure la ligne "Total" si elle existe
data_pannes_filtered = data_pannes[data_pannes['Année'] != 'Total']

# Calculer les fréquences des nombres de pannes
unique_pannes = data_pannes_filtered['Nombre de pannes'].value_counts().sort_index()

# Calcul des probabilités
probabilities = unique_pannes / unique_pannes.sum()

# Création de l'histogramme des probabilités
plt.figure(figsize=(10, 6))
plt.bar(unique_pannes.index, probabilities, color='skyblue', edgecolor='black', alpha=0.7)
plt.title("Distribution des probabilités des pannes de serveur", fontsize=14)
plt.xlabel("Nombre de pannes", fontsize=12)
plt.ylabel("Probabilité", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(unique_pannes.index)
plt.show()
```
